## Handling exceptions with try / except / finally

```python
try:
    result = 5 / 0
except ZeroDivisionError as e:
    print(f"Caught an exception: {e}")
finally:
    print("This block is executed regardless of whether an exception occurred or not.")
```

## Multiple except Blocks

```python
try:
    x = int("hello")
except ValueError:
    print("That's not an integer!")
except TypeError:
    print("Wrong type!")
```

## Catching Everything (not always recommended)
### Use this for debugging or generic logging, but don't overuse it — catching Exception hides bugs.

```python
try:
    x = 5 / 0
except Exception as e:
    print(f"An error occurred: {e}")
```

## Understanding the stack trace

```python
def level_one():
    level_two()

def level_two():
    raise RuntimeError("Something bad happened in level two!")

level_one()
```
