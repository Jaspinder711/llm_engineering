## Tuple, List and Set

```python
## Tuple is immuatable, meaning once created, its elements cannot be changed, added, or removed. Tuples are defined using parentheses ().
tuple_example = (1, 2, 3)
type(tuple_example)  # Output: <class 'tuple'>
tuple_example
```

```python
## List is mutable, meaning its elements can be changed, added, or removed. Lists are defined using square brackets [].
list_example = [1, 2, 3]
type(list_example)  # Output: <class 'list'>
list_example.append(4)
list_example
```

```python
# Can't add duplicates in a set. Sets are defined using curly braces {}.
set_example = {1, 2, 3}
type(set_example)  # Output: <class 'set'>
set_example.add(2)  # Adding a duplicate value
set_example
```

## Dictionary

```python
person = {
    "name" : "Alice",
    "age" : 25,
    "city" : "New York"
}

print(person["name"])
print(person.get("age"))

person["job"] = "Engineer"
print(person)

## changing a value
person["city"] = "Los Angeles"
print(person)

## default value if key doesn't exist
print(person.get("country", "USA"))
## didn't make change in dictionary
print(person)
```

## Classes

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def description(self):
        return f"{self.year} {self.make} {self.model}"

    def start_engine(self):
        return "Vroom!"
```

```python
my_car = Car("Toyota", "Camry", 2020)
print(my_car.description())
print(my_car.start_engine())
```

## Pickling
### Pickling is a way to serialize (convert) a Python object into a binary format so that it can be saved to a file and later reloaded. This is especially useful when you want to save complex objects like lists, dictionaries, or instances of classes to a file.



```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."
```

```python
## Pickle an object, here is how to save an instance of the Person class to a file using the pickle module:
import pickle

person = Person("Alice", 30)

with open("person.pkl", "wb") as file:
    pickle.dump(person,file)

print("Person object has been pickled and saved to person.pkl")
```

```python
## Unpickle the object, here is how to load the pickled Person object from the file:
with open("person.pkl", "rb") as file:
    loaded_person = pickle.load(file)

print(f"Loaded Person: {loaded_person.name}, Age: {loaded_person.age}")
```
